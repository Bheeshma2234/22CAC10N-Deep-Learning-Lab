```python
# ============================================================
# EXPERIMENT: CNN FOR MNIST IMAGE CLASSIFICATION
# ============================================================

# ============================================================
# 1. IMPORT PACKAGES
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf
import time

from tensorflow.keras import Sequential
from tensorflow.keras.layers import (
    Conv2D, MaxPooling2D, BatchNormalization,
    Flatten, Dense, Dropout
)
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_curve,
    auc
)
from sklearn.preprocessing import label_binarize

print("TensorFlow Version:", tf.__version__)


# ============================================================
# 2. LOAD MNIST DATASET
# ============================================================

(x_train, y_train), (x_test, y_test) = \
    tf.keras.datasets.mnist.load_data()

# Train - Validation Split
x_train, x_val, y_train, y_val = train_test_split(
    x_train,
    y_train,
    test_size=0.10,
    random_state=42,
    stratify=y_train
)

# Add channel dimension
x_train = x_train[..., np.newaxis]
x_val = x_val[..., np.newaxis]
x_test = x_test[..., np.newaxis]

# Normalize
x_train = x_train.astype("float32") / 255.0
x_val = x_val.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

print("\nDataset Loaded Successfully")


# ============================================================
# 3. VISUALIZE DATASET - ONE SAMPLE PER CLASS
# ============================================================

plt.figure(figsize=(12, 5))

for digit in range(10):

    index = np.where(y_train == digit)[0][0]

    plt.subplot(2, 5, digit + 1)
    plt.imshow(
        x_train[index].squeeze(),
        cmap="gray"
    )
    plt.title(f"Class {digit}")
    plt.axis("off")

plt.suptitle("One Sample Per Class")
plt.tight_layout()
plt.show()


# ============================================================
# 4. CLASS-WISE DISTRIBUTION - TRAINING SET
# ============================================================

class_count = (
    pd.Series(y_train)
    .value_counts()
    .sort_index()
)

distribution = pd.DataFrame({
    "Class": class_count.index,
    "Training Samples": class_count.values
})

print("\nClass-wise Distribution:")
print(
    distribution.to_string(index=False)
)

plt.figure(figsize=(8, 5))

plt.bar(
    distribution["Class"],
    distribution["Training Samples"]
)

plt.xlabel("Digit Class")
plt.ylabel("Number of Samples")
plt.title("Class-wise Distribution - Training Set")
plt.xticks(range(10))
plt.grid(axis="y")

plt.show()


# ============================================================
# 5. TRAIN / VALIDATION / TEST SIZES AND SHAPES
# ============================================================

print("\nDataset Sizes:")
print("Training   :", x_train.shape[0])
print("Validation :", x_val.shape[0])
print("Testing    :", x_test.shape[0])

print("\nDataset Shapes:")
print("X Train:", x_train.shape)
print("Y Train:", y_train.shape)
print("X Val  :", x_val.shape)
print("Y Val  :", y_val.shape)
print("X Test :", x_test.shape)
print("Y Test :", y_test.shape)


# ============================================================
# 6. CNN ARCHITECTURE
#
# C1 -> P1 -> C2 -> P2 -> C3
# -> Flatten -> FC1(256)
# -> FC2(128) -> Output(10)
#
# Filters:
# C1 = 32
# C2 = 64
# C3 = 128
# ============================================================

model = Sequential([

    # ---------------- C1 ----------------
    Conv2D(
        32,
        (3, 3),
        activation="relu",
        padding="same",
        input_shape=(28, 28, 1)
    ),
    BatchNormalization(),

    # ---------------- P1 ----------------
    MaxPooling2D((2, 2)),

    # ---------------- C2 ----------------
    Conv2D(
        64,
        (3, 3),
        activation="relu",
        padding="same"
    ),
    BatchNormalization(),

    # ---------------- P2 ----------------
    MaxPooling2D((2, 2)),

    # ---------------- C3 ----------------
    Conv2D(
        128,
        (3, 3),
        activation="relu",
        padding="same"
    ),
    BatchNormalization(),

    # ---------------- Flatten ----------------
    Flatten(),

    # ---------------- FC1 ----------------
    Dense(
        256,
        activation="relu"
    ),
    Dropout(0.3),

    # ---------------- FC2 ----------------
    Dense(
        128,
        activation="relu"
    ),
    Dropout(0.3),

    # ---------------- Output ----------------
    Dense(
        10,
        activation="softmax"
    )
])


# ============================================================
# 7. MODEL SUMMARY IN KERAS TABULAR FORMAT
# ============================================================

model.build((None, 28, 28, 1))

print("\n")
print("=" * 80)
print("COMMON MODEL ARCHITECTURE")
print("=" * 80)

# Keras summary gives the table format
# similar to:
# Layer (type) | Output Shape | Param #

model.summary()


# ============================================================
# ADDITIONAL MODEL SUMMARY IN DATAFRAME FORMAT
# ============================================================

summary_table = []

for i, layer in enumerate(model.layers):

    try:
        output_shape = str(layer.output.shape)
    except:
        output_shape = "N/A"

    summary_table.append([
        i + 1,
        layer.name,
        layer.__class__.__name__,
        output_shape,
        layer.count_params()
    ])

model_table = pd.DataFrame(
    summary_table,
    columns=[
        "No.",
        "Layer",
        "Type",
        "Output Shape",
        "Parameters"
    ]
)

print("\nMODEL SUMMARY TABLE:")
print(
    model_table.to_string(index=False)
)

print(
    "\nTotal Parameters:",
    model.count_params()
)


# ============================================================
# 8. COMPILE AND TRAIN
#
# Early Stopping
# Checkpoint
# Maximum Epochs = 20
# Patience = 5
# Learning Rate = 0.001
# ============================================================

learning_rate = 0.001
patience = 5
epochs = 20

model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=learning_rate
    ),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)


# ============================================================
# EARLY STOPPING
# ============================================================

early_stop = EarlyStopping(
    monitor="val_loss",
    patience=patience,
    restore_best_weights=True,
    verbose=1
)


# ============================================================
# HDF5 CHECKPOINT
# ============================================================

checkpoint = ModelCheckpoint(
    "best_cnn_mnist.h5",
    monitor="val_loss",
    save_best_only=True,
    save_weights_only=False,
    verbose=1
)


# ============================================================
# TRAINING
# ============================================================

print("\n")
print("=" * 80)
print("TRAINING CNN")
print("=" * 80)

start_time = time.time()

history = model.fit(
    x_train,
    y_train,
    validation_data=(x_val, y_val),
    epochs=epochs,
    batch_size=128,
    callbacks=[
        early_stop,
        checkpoint
    ],
    verbose=1
)

end_time = time.time()

training_time = end_time - start_time


# ============================================================
# SAVE BEST MODEL AS HDF5
# ============================================================

# EarlyStopping restored the best weights.
# Save those restored weights as the final best HDF5 model.

model.save(
    "best_cnn_mnist.h5"
)

print("\nBest model saved as:")
print("best_cnn_mnist.h5")


# ============================================================
# BEST EPOCH AND ACTUAL EPOCHS
# ============================================================

actual_epochs = len(
    history.history["loss"]
)

best_epoch = (
    np.argmin(
        history.history["val_loss"]
    ) + 1
)

print("\nBest Epoch:", best_epoch)
print("Actual Epochs Run:", actual_epochs)


# ============================================================
# 9. ACCURACY AND LOSS GRAPHS
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    history.history["accuracy"],
    label="Training Accuracy"
)

plt.plot(
    history.history["val_accuracy"],
    label="Validation Accuracy"
)

plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("Training vs Validation Accuracy")

plt.legend()
plt.grid()

plt.show()


# ============================================================
# LOSS GRAPH
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    history.history["loss"],
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Training vs Validation Loss")

plt.legend()
plt.grid()

plt.show()


# ============================================================
# 10. PERFORMANCE MATRIX
# ============================================================

y_prob = model.predict(
    x_test,
    verbose=0
)

y_pred = np.argmax(
    y_prob,
    axis=1
)


# ============================================================
# PERFORMANCE METRICS
# ============================================================

accuracy = accuracy_score(
    y_test,
    y_pred
)

precision = precision_score(
    y_test,
    y_pred,
    average="weighted"
)

recall = recall_score(
    y_test,
    y_pred,
    average="weighted"
)

f1 = f1_score(
    y_test,
    y_pred,
    average="weighted"
)


# ============================================================
# PRINT EXPERIMENT SUMMARY
# ============================================================

print("\n")
print("-" * 70)
print("TECHNIQUE : CNN")
print("-" * 70)

print(
    f"Restoring model weights from the end of "
    f"the best epoch: {best_epoch}."
)

print()

print(
    f"Accuracy : {accuracy * 100:.2f} %"
)

print(
    f"Precision : {precision:.4f}"
)

print(
    f"Recall : {recall:.4f}"
)

print(
    f"F1 Score : {f1:.4f}"
)

print(
    f"Actual Epochs Run : {actual_epochs}"
)

print(
    f"Early Stopping Patience : {patience}"
)

print(
    f"Learning Rate : {learning_rate}"
)

print(
    "Conv Layer 1 : 32 Filters - 3x3 - ReLU"
)

print(
    "Conv Layer 2 : 64 Filters - 3x3 - ReLU"
)

print(
    "Conv Layer 3 : 128 Filters - 3x3 - ReLU"
)

print(
    "Hidden Layer 1 : 256 - ReLU"
)

print(
    "Hidden Layer 2 : 128 - ReLU"
)

print(
    "Output Layer : 10 - Softmax"
)

print(
    f"Time : {training_time:.2f} seconds"
)


# ============================================================
# CLASSIFICATION REPORT
# INCLUDING SUPPORT
# ============================================================

print("\nClassification Report:")

report = classification_report(
    y_test,
    y_pred,
    target_names=[
        str(i) for i in range(10)
    ],
    output_dict=True
)

report_df = pd.DataFrame(
    report
).transpose()

print(
    report_df.round(4).to_string()
)


# ============================================================
# CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    y_test,
    y_pred
)

print("\nConfusion Matrix:")
print(cm)

plt.figure(figsize=(8, 7))

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=range(10)
)

disp.plot(
    cmap="Blues",
    values_format="d"
)

plt.title("CNN Confusion Matrix")
plt.show()


# ============================================================
# 11. ROC CURVE
# ONE-VS-REST
# ============================================================

# Convert labels to binary form
y_test_bin = label_binarize(
    y_test,
    classes=range(10)
)

fpr = {}
tpr = {}
roc_auc = {}


# ============================================================
# ONE-VS-REST ROC
# ============================================================

for i in range(10):

    fpr[i], tpr[i], _ = roc_curve(
        y_test_bin[:, i],
        y_prob[:, i]
    )

    roc_auc[i] = auc(
        fpr[i],
        tpr[i]
    )


# ============================================================
# MICRO-AVERAGE ROC
# ============================================================

fpr_micro, tpr_micro, _ = roc_curve(
    y_test_bin.ravel(),
    y_prob.ravel()
)

micro_auc = auc(
    fpr_micro,
    tpr_micro
)


# ============================================================
# PRINT ONE-VS-REST AUC VALUES
# ============================================================

print("\n")
print("=" * 70)
print("ROC-AUC RESULTS - ONE VS REST")
print("=" * 70)

for i in range(10):

    print(
        f"Class {i} AUC : "
        f"{roc_auc[i]:.4f}"
    )


print(
    f"\nMicro Average AUC : {micro_auc:.4f}"
)


# ============================================================
# ROC CURVE - ONE VS REST + MICRO AVERAGE
# ============================================================

plt.figure(figsize=(10, 7))

for i in range(10):

    plt.plot(
        fpr[i],
        tpr[i],
        label=(
            f"Class {i} "
            f"(AUC = {roc_auc[i]:.3f})"
        )
    )


# Micro Average
plt.plot(
    fpr_micro,
    tpr_micro,
    linewidth=3,
    label=(
        f"Micro Average "
        f"(AUC = {micro_auc:.3f})"
    )
)


# Random classifier
plt.plot(
    [0, 1],
    [0, 1],
    "k--",
    label="Random"
)

plt.xlabel(
    "False Positive Rate"
)

plt.ylabel(
    "True Positive Rate"
)

plt.title(
    "ROC Curve - One vs Rest + Micro Average"
)

plt.legend(
    loc="lower right",
    fontsize=8
)

plt.grid()

plt.show()


# ============================================================
# 12. PERFORMANCE SUMMARY IN TABULAR FORM
# ============================================================

performance = pd.DataFrame({

    "Metric": [
        "Accuracy",
        "Precision",
        "Recall",
        "F1-Score",
        "Micro ROC-AUC"
    ],

    "Score": [
        accuracy,
        precision,
        recall,
        f1,
        micro_auc
    ]
})

print("\n")
print("=" * 70)
print("PERFORMANCE SUMMARY")
print("=" * 70)

print(
    performance.to_string(
        index=False,
        formatters={
            "Score": "{:.4f}".format
        }
    )
)


# ============================================================
# FINAL INFORMATION
# ============================================================

print("\n")
print("=" * 70)
print("CNN EXPERIMENT COMPLETED SUCCESSFULLY")
print("=" * 70)

print(
    "\nBest HDF5 Model : best_cnn_mnist.h5"
)

print(
    f"Best Epoch : {best_epoch}"
)

print(
    f"Actual Epochs Run : {actual_epochs}"
)

print(
    f"Training Time : {training_time:.2f} seconds"
)

print(
    f"Test Accuracy : {accuracy * 100:.2f}%"
)

print(
    f"Micro ROC-AUC : {micro_auc:.4f}"
)

print("=" * 70)
```
