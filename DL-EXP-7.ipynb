```python
# ============================================================
# DL EXPERIMENT 7
# LSTM FOR NEXT-WORD PREDICTION
# Dataset: 3 Project Gutenberg Books
# ============================================================

import re
import random
import warnings
import numpy as np
import matplotlib.pyplot as plt
import requests
import nltk
import tensorflow as tf

from collections import Counter
from nltk import word_tokenize, pos_tag
from nltk.corpus import wordnet
from nltk.translate.bleu_score import corpus_bleu, SmoothingFunction
from nltk.translate.meteor_score import meteor_score
from nltk.metrics import edit_distance

from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

warnings.filterwarnings("ignore")


# ------------------------------------------------------------
# 1. SET RANDOM SEEDS
# ------------------------------------------------------------

np.random.seed(42)
random.seed(42)
tf.random.set_seed(42)

print("=" * 60)
print("LSTM NEXT-WORD PREDICTION")
print("=" * 60)


# ------------------------------------------------------------
# 2. DOWNLOAD NLTK PACKAGES
# ------------------------------------------------------------

for package in [
    "punkt",
    "punkt_tab",
    "averaged_perceptron_tagger_eng",
    "wordnet"
]:
    nltk.download(package, quiet=True)


# ------------------------------------------------------------
# 3. DOWNLOAD THREE BOOKS
# ------------------------------------------------------------

book_ids = {
    11: "Alice's Adventures in Wonderland",
    84: "Frankenstein",
    1661: "The Adventures of Sherlock Holmes"
}

all_books = []
loaded_books = []

print("\nDOWNLOADING DATASET")
print("-" * 50)

for book_id, book_title in book_ids.items():

    try:

        url = (
            f"https://www.gutenberg.org/cache/epub/"
            f"{book_id}/pg{book_id}.txt"
        )

        response = requests.get(url, timeout=60)
        response.raise_for_status()

        book_text = response.text

        start = re.search(
            r"\*\*\* START OF (?:THE|THIS) PROJECT GUTENBERG EBOOK.*?\*\*\*",
            book_text,
            flags=re.I | re.S
        )

        end = re.search(
            r"\*\*\* END OF (?:THE|THIS) PROJECT GUTENBERG EBOOK",
            book_text,
            flags=re.I
        )

        if start:
            book_text = book_text[start.end():]

        if end:
            book_text = re.split(
                r"\*\*\* END OF (?:THE|THIS) PROJECT GUTENBERG EBOOK",
                book_text,
                flags=re.I
            )[0]

        all_books.append(book_text)
        loaded_books.append(book_title)

        print("Loaded:", book_title)

    except Exception as e:

        print("Skipped:", book_title, "|", e)


if not all_books:

    raise RuntimeError(
        "Books could not be downloaded. Check your internet connection."
    )


raw_text = "\n".join(all_books)

print("\nBOOKS LOADED:", len(loaded_books))
print("BOOK TITLES:", loaded_books)
print("TOTAL CHARACTERS:", len(raw_text))

print("\nORIGINAL TEXT:\n", raw_text[:200])


# ------------------------------------------------------------
# 4. TEXT PREPROCESSING AND CLEANING
# ------------------------------------------------------------

text = raw_text.lower()

text = re.sub(
    r"[^a-z\s']",
    " ",
    text
)

text = re.sub(
    r"\s+",
    " ",
    text
).strip()

print("\nCLEANED TEXT")
print(text[:200])


tokens = word_tokenize(text)

print("\nTOTAL WORD TOKENS:", len(tokens))
print("SAMPLE TOKENS:", tokens[:15])


# WordNet dictionary analysis

sample_words = [
    w for w in tokens[:50]
    if w.isalpha()
]

dictionary_words = [
    w for w in sample_words
    if wordnet.synsets(w)
]

unknown_words = [
    w for w in sample_words
    if not wordnet.synsets(w)
]

print("\nWORDNET ANALYSIS")

print(
    "Dictionary words:",
    dictionary_words[:10]
)

print(
    "Words not found:",
    unknown_words[:10]
)


# Part-of-speech tagging

print("\nPART-OF-SPEECH TAGGING")

print(
    pos_tag(tokens[:15])
)


# ------------------------------------------------------------
# 5. TOKENIZATION AND WORD EMBEDDINGS
# ------------------------------------------------------------

MAX_WORDS = 5000
SEQ_LEN = 8
EMBED_DIM = 32


tokenizer = Tokenizer(
    num_words=MAX_WORDS,
    oov_token="<OOV>"
)

tokenizer.fit_on_texts([text])


vocab_size = min(
    MAX_WORDS,
    len(tokenizer.word_index) + 1
)


word_ids = tokenizer.texts_to_sequences([text])[0]


print("\nTOKENIZATION DETAILS")

print(
    "Vocabulary size:",
    vocab_size
)

print(
    "Sequence length:",
    SEQ_LEN
)

print(
    "Embedding dimensions:",
    EMBED_DIM
)


print("\nWORD-TO-INDEX MAPPING")

print(
    list(tokenizer.word_index.items())[:15]
)


# ------------------------------------------------------------
# 6. CREATE SEQUENCES
# ------------------------------------------------------------

n = len(word_ids)

split1 = int(n * 0.80)
split2 = int(n * 0.90)


train_ids = word_ids[:split1]

val_ids = word_ids[
    split1:split2
]

test_ids = word_ids[
    split2:
]


def create_sequences(ids, seq_len):

    X = []
    y = []

    for i in range(seq_len, len(ids)):

        X.append(
            ids[i - seq_len:i]
        )

        y.append(
            ids[i]
        )

    return (
        np.asarray(X, dtype=np.int32),
        np.asarray(y, dtype=np.int32)
    )


X_train, y_train = create_sequences(
    train_ids,
    SEQ_LEN
)

X_val, y_val = create_sequences(
    val_ids,
    SEQ_LEN
)

X_test, y_test = create_sequences(
    test_ids,
    SEQ_LEN
)


print("\nSEQUENCE DATA")

print(
    "X_train:",
    X_train.shape
)

print(
    "y_train:",
    y_train.shape
)

print(
    "X_val:",
    X_val.shape
)

print(
    "y_val:",
    y_val.shape
)

print(
    "X_test:",
    X_test.shape
)

print(
    "y_test:",
    y_test.shape
)


id_to_word = {
    index: word
    for word, index in tokenizer.word_index.items()
    if index < vocab_size
}


def decode_sequence(sequence):

    return [
        id_to_word.get(
            int(i),
            "?"
        )
        for i in sequence
    ]


print(
    "\nSAMPLE INPUT WORDS:",
    decode_sequence(X_train[0])
)

print(
    "INPUT TOKEN IDS:",
    X_train[0]
)

print(
    "ACTUAL NEXT WORD:",
    id_to_word.get(
        int(y_train[0]),
        "?"
    )
)


# ------------------------------------------------------------
# 7. BUILD AND COMPILE THE LSTM MODEL
# ------------------------------------------------------------

model = Sequential([

    tf.keras.Input(
        shape=(SEQ_LEN,)
    ),

    Embedding(
        input_dim=vocab_size,
        output_dim=EMBED_DIM,
        mask_zero=True,
        name="Word_Embedding"
    ),

    LSTM(
        32,
        name="LSTM_Layer"
    ),

    Dropout(0.2),

    Dense(
        32,
        activation="relu"
    ),

    Dense(
        vocab_size,
        activation="softmax",
        name="Next_Word_Output"
    )
])


# Accuracy metrics are intentionally removed.
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy"
)


print("\nMODEL SUMMARY")

model.summary()


# ------------------------------------------------------------
# 8. TRAIN THE MODEL
# ------------------------------------------------------------

early_stop = EarlyStopping(
    monitor="val_loss",
    patience=3,
    restore_best_weights=True
)


print("\nTRAINING LSTM MODEL")

print("Maximum epochs: 10")
print("Early stopping patience: 3")


history = model.fit(

    X_train,
    y_train,

    validation_data=(
        X_val,
        y_val
    ),

    epochs=10,

    batch_size=128,

    callbacks=[
        early_stop
    ],

    shuffle=False,

    verbose=1
)


print("\nTraining completed.")

print(
    "Epochs completed:",
    len(history.history["loss"])
)


# ------------------------------------------------------------
# 9. NLP PERFORMANCE METRICS
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("NLP PERFORMANCE METRICS")
print("=" * 60)


# Only 2,000 samples are used for metric calculation
# to make execution faster.

METRIC_SAMPLES = min(
    2000,
    len(X_test)
)


print(
    "Metric samples:",
    METRIC_SAMPLES
)


X_metric = X_test[
    :METRIC_SAMPLES
]

y_metric = y_test[
    :METRIC_SAMPLES
]


print("\nGENERATING TEST-SET PREDICTIONS...")


metric_probabilities = model.predict(
    X_metric,
    batch_size=128,
    verbose=0
)


predicted_ids = np.argmax(
    metric_probabilities,
    axis=1
)


actual_words = [
    id_to_word.get(
        int(word_id),
        "<OOV>"
    )
    for word_id in y_metric
]


predicted_words = [
    id_to_word.get(
        int(word_id),
        "<OOV>"
    )
    for word_id in predicted_ids
]


# Free large prediction array
del metric_probabilities


print(
    "Actual words:",
    len(actual_words)
)

print(
    "Predicted words:",
    len(predicted_words)
)


# ------------------------------------------------------------
# 9.1 BLEU SCORES
# ------------------------------------------------------------

smooth = SmoothingFunction().method1


# Correct corpus BLEU structure:
# references = [[reference_sentence]]
# hypotheses = [hypothesis_sentence]

references = [
    actual_words
]

hypotheses = [
    predicted_words
]


bleu1 = corpus_bleu(
    [references],
    hypotheses,
    weights=(1, 0, 0, 0),
    smoothing_function=smooth
)


bleu2 = corpus_bleu(
    [references],
    hypotheses,
    weights=(0.5, 0.5, 0, 0),
    smoothing_function=smooth
)


bleu3 = corpus_bleu(
    [references],
    hypotheses,
    weights=(1 / 3, 1 / 3, 1 / 3, 0),
    smoothing_function=smooth
)


# ------------------------------------------------------------
# 9.2 METEOR
# ------------------------------------------------------------

meteor = meteor_score(
    [actual_words],
    predicted_words
)


# ------------------------------------------------------------
# 9.3 ROUGE RECALL
# ------------------------------------------------------------

def get_ngrams(words, n):

    return [
        tuple(words[i:i + n])
        for i in range(
            len(words) - n + 1
        )
    ]


def rouge_recall(
    reference,
    hypothesis,
    n
):

    reference_ngrams = Counter(
        get_ngrams(
            reference,
            n
        )
    )

    hypothesis_ngrams = Counter(
        get_ngrams(
            hypothesis,
            n
        )
    )

    overlap = sum(
        (reference_ngrams & hypothesis_ngrams).values()
    )

    total = sum(
        reference_ngrams.values()
    )

    if total == 0:
        return 0.0

    return overlap / total


rouge1 = rouge_recall(
    actual_words,
    predicted_words,
    1
)


rouge2 = rouge_recall(
    actual_words,
    predicted_words,
    2
)


rouge3 = rouge_recall(
    actual_words,
    predicted_words,
    3
)


# ------------------------------------------------------------
# 9.4 WORD ERROR RATE
# ------------------------------------------------------------

word_errors = edit_distance(
    actual_words,
    predicted_words
)


wer = (
    word_errors /
    max(len(actual_words), 1)
)


# ------------------------------------------------------------
# 9.5 CHARACTER ERROR RATE
# ------------------------------------------------------------
# Word-wise character error calculation is used
# to avoid extremely large character-level matrices.

character_errors = 0
total_characters = 0


for actual, predicted in zip(
    actual_words,
    predicted_words
):

    character_errors += edit_distance(
        actual,
        predicted
    )

    total_characters += len(actual)


cer = (
    character_errors /
    max(total_characters, 1)
)


# ------------------------------------------------------------
# 9.6 DISPLAY ALL METRICS
# ------------------------------------------------------------

print("\nFINAL NLP METRICS")
print("-" * 50)

print(
    f"BLEU-1              : {bleu1:.4f}"
)

print(
    f"BLEU-2              : {bleu2:.4f}"
)

print(
    f"BLEU-3              : {bleu3:.4f}"
)

print(
    f"METEOR              : {meteor:.4f}"
)

print(
    f"ROUGE-1 Recall      : {rouge1:.4f}"
)

print(
    f"ROUGE-2 Recall      : {rouge2:.4f}"
)

print(
    f"ROUGE-3 Recall      : {rouge3:.4f}"
)

print(
    f"WER                 : {wer:.4f}"
)

print(
    f"CER                 : {cer:.4f}"
)


# ------------------------------------------------------------
# 10. BAR GRAPH - QUALITY METRICS
# ------------------------------------------------------------

quality_names = [
    "BLEU-1",
    "BLEU-2",
    "BLEU-3",
    "METEOR",
    "ROUGE-1",
    "ROUGE-2",
    "ROUGE-3"
]


quality_values = [
    bleu1,
    bleu2,
    bleu3,
    meteor,
    rouge1,
    rouge2,
    rouge3
]


plt.figure(
    figsize=(11, 6)
)


bars = plt.bar(
    quality_names,
    quality_values
)


plt.title(
    "NLP Quality Metrics"
)

plt.xlabel(
    "Metrics"
)

plt.ylabel(
    "Score"
)


plt.ylim(
    0,
    max(
        1.0,
        max(quality_values) * 1.15
    )
)


for bar, value in zip(
    bars,
    quality_values
):

    plt.text(
        bar.get_x() +
        bar.get_width() / 2,

        bar.get_height() + 0.01,

        f"{value:.3f}",

        ha="center"
    )


plt.xticks(
    rotation=20
)

plt.tight_layout()

plt.show()


# ------------------------------------------------------------
# 11. BAR GRAPH - ERROR METRICS
# ------------------------------------------------------------

error_names = [
    "WER",
    "CER"
]


error_values = [
    wer,
    cer
]


plt.figure(
    figsize=(7, 5)
)


bars = plt.bar(
    error_names,
    error_values
)


plt.title(
    "Error Metrics"
)

plt.xlabel(
    "Metrics"
)

plt.ylabel(
    "Error Rate"
)


for bar, value in zip(
    bars,
    error_values
):

    plt.text(
        bar.get_x() +
        bar.get_width() / 2,

        bar.get_height() + 0.01,

        f"{value:.3f}",

        ha="center"
    )


plt.tight_layout()

plt.show()


# ------------------------------------------------------------
# 12. WORD EMBEDDING VECTORS
# ------------------------------------------------------------

embedding_matrix = model.get_layer(
    "Word_Embedding"
).get_weights()[0]


print("\nWORD EMBEDDING MATRIX")

print(
    "Embedding matrix shape:",
    embedding_matrix.shape
)


for word in [
    "alice",
    "rabbit",
    "queen",
    "king",
    "said"
]:

    word_id = tokenizer.word_index.get(
        word
    )

    if (
        word_id is not None
        and word_id < vocab_size
    ):

        print("\nWord:", word)

        print(
            "Token ID:",
            word_id
        )

        print(
            "First 10 vector values:",
            np.round(
                embedding_matrix[word_id][:10],
                4
            )
        )


print("\nFIRST FIVE VECTORS")

print(
    np.round(
        embedding_matrix[1:6, :8],
        4
    )
)


# ------------------------------------------------------------
# 13. NEXT-WORD PREDICTION
# ------------------------------------------------------------

def predict_next_word(
    input_text,
    top_n=5,
    check_answer=False
):

    cleaned = input_text.lower()

    cleaned = re.sub(
        r"[^a-z\s']",
        " ",
        cleaned
    )

    cleaned = re.sub(
        r"\s+",
        " ",
        cleaned
    ).strip()


    sequence = tokenizer.texts_to_sequences(
        [cleaned]
    )[0]


    sequence = sequence[-SEQ_LEN:]


    if not sequence:

        print(
            "Please enter valid words."
        )

        return


    if all(
        word_id == 1
        for word_id in sequence
    ):

        print(
            "Input words are outside "
            "the training vocabulary."
        )

        print(
            "Try words used in the training books."
        )

        return


    padded = pad_sequences(
        [sequence],
        maxlen=SEQ_LEN,
        padding="pre"
    )


    probabilities = model.predict(
        padded,
        verbose=0
    )[0].copy()


    # Exclude padding and OOV tokens

    probabilities[0] = 0
    probabilities[1] = 0


    valid_ids = [

        i

        for i in range(
            2,
            len(probabilities)
        )

        if i in id_to_word
    ]


    top_ids = sorted(
        valid_ids,

        key=lambda i:
        probabilities[i],

        reverse=True
    )[:top_n]


    if not top_ids:

        print(
            "No valid prediction found."
        )

        return


    print("\nINPUT:", input_text)

    print(
        "ENCODED INPUT:",
        padded[0]
    )


    print(
        "\nTOP NEXT-WORD PREDICTIONS"
    )

    print(
        "-" * 45
    )


    for rank, word_id in enumerate(
        top_ids,
        start=1
    ):

        word = id_to_word[word_id]

        probability = (
            probabilities[word_id] * 100
        )


        print(
            f"{rank}. {word:<18}"
            f"Probability: "
            f"{probability:.2f}%"
        )


    predicted_word = id_to_word[
        top_ids[0]
    ]


    print(
        "\nPREDICTED NEXT WORD:",
        predicted_word
    )


    print(
        "COMPLETED TEXT:",
        input_text + " " + predicted_word
    )


    # --------------------------------------------------------
    # CHECK PREDICTION
    # --------------------------------------------------------

    if check_answer:

        actual_word = input(
            "\nEnter the actual next word "
            "(press Enter to skip): "
        ).strip()


        if actual_word:

            actual_word = re.sub(
                r"[^a-z']",
                "",
                actual_word.lower()
            )


            print(
                "\nCORRECTNESS CHECK"
            )

            print(
                "Predicted word:",
                predicted_word
            )

            print(
                "Actual word:",
                actual_word
            )


            if (
                predicted_word.lower()
                == actual_word
            ):

                print(
                    "RESULT: CORRECT"
                )

            else:

                print(
                    "RESULT: WRONG"
                )

        else:

            print(
                "Correctness check skipped."
            )


# ------------------------------------------------------------
# 14. SAMPLE PREDICTIONS
# ------------------------------------------------------------

print(
    "\nSAMPLE PREDICTIONS"
)


predict_next_word(
    "alice was"
)


predict_next_word(
    "the white rabbit"
)


predict_next_word(
    "the queen"
)


# ------------------------------------------------------------
# 15. INTERACTIVE USER INPUT
# Type exit to stop
# ------------------------------------------------------------

print(
    "\n" + "=" * 60
)

print(
    "INTERACTIVE NEXT-WORD PREDICTION"
)

print(
    "Enter a sentence or type 'exit' to stop."
)

print(
    "=" * 60
)


while True:

    user_text = input(
        "\nEnter a sentence: "
    ).strip()


    if user_text.lower() == "exit":

        print(
            "Prediction stopped. "
            "Experiment completed!"
        )

        break


    if not user_text:

        print(
            "Please enter a sentence."
        )

        continue


    predict_next_word(
        user_text,
        top_n=5,
        check_answer=True
    )


print(
    "\nDL EXPERIMENT 7 FINISHED SUCCESSFULLY"
)
```

```python
# ============================================================
# PERFORMANCE METRICS + BAR GRAPHS
# ============================================================

import numpy as np
import matplotlib.pyplot as plt

from collections import Counter
from nltk.translate.bleu_score import corpus_bleu, SmoothingFunction
from nltk.translate.meteor_score import meteor_score
from nltk.metrics import edit_distance


# ------------------------------------------------------------
# 1. SELECT TEST SAMPLES
# ------------------------------------------------------------

# Use only 2,000 samples for faster evaluation
METRIC_SAMPLES = min(2000, len(X_test))

X_metric = X_test[:METRIC_SAMPLES]
y_metric = y_test[:METRIC_SAMPLES]

print("=" * 60)
print("PERFORMANCE METRICS")
print("=" * 60)

print("Number of test samples:", METRIC_SAMPLES)


# ------------------------------------------------------------
# 2. PREDICT NEXT WORDS
# ------------------------------------------------------------

print("\nGenerating predictions...")

predictions = model.predict(
    X_metric,
    batch_size=128,
    verbose=0
)

predicted_ids = np.argmax(
    predictions,
    axis=1
)

del predictions


# Convert token IDs to words

actual_words = [
    id_to_word.get(
        int(word_id),
        "<OOV>"
    )
    for word_id in y_metric
]

predicted_words = [
    id_to_word.get(
        int(word_id),
        "<OOV>"
    )
    for word_id in predicted_ids
]


print("Predictions generated successfully.")


# ------------------------------------------------------------
# 3. BLEU-1, BLEU-2, BLEU-3
# ------------------------------------------------------------

smooth = SmoothingFunction().method1

references = [actual_words]
hypotheses = [predicted_words]


bleu1 = corpus_bleu(
    [references],
    hypotheses,
    weights=(1, 0, 0, 0),
    smoothing_function=smooth
)


bleu2 = corpus_bleu(
    [references],
    hypotheses,
    weights=(0.5, 0.5, 0, 0),
    smoothing_function=smooth
)


bleu3 = corpus_bleu(
    [references],
    hypotheses,
    weights=(1/3, 1/3, 1/3, 0),
    smoothing_function=smooth
)


# ------------------------------------------------------------
# 4. METEOR
# ------------------------------------------------------------

meteor = meteor_score(
    [actual_words],
    predicted_words
)


# ------------------------------------------------------------
# 5. ROUGE-1, ROUGE-2, ROUGE-3 RECALL
# ------------------------------------------------------------

def get_ngrams(words, n):

    return [
        tuple(words[i:i+n])
        for i in range(
            len(words) - n + 1
        )
    ]


def rouge_recall(reference, hypothesis, n):

    reference_ngrams = Counter(
        get_ngrams(reference, n)
    )

    hypothesis_ngrams = Counter(
        get_ngrams(hypothesis, n)
    )

    overlap = sum(
        (reference_ngrams & hypothesis_ngrams).values()
    )

    total = sum(
        reference_ngrams.values()
    )

    if total == 0:
        return 0.0

    return overlap / total


rouge1 = rouge_recall(
    actual_words,
    predicted_words,
    1
)


rouge2 = rouge_recall(
    actual_words,
    predicted_words,
    2
)


rouge3 = rouge_recall(
    actual_words,
    predicted_words,
    3
)


# ------------------------------------------------------------
# 6. WORD ERROR RATE (WER)
# ------------------------------------------------------------

word_errors = edit_distance(
    actual_words,
    predicted_words
)

wer = word_errors / max(
    len(actual_words),
    1
)


# ------------------------------------------------------------
# 7. CHARACTER ERROR RATE (CER)
# ------------------------------------------------------------

# Word-wise calculation is used for faster execution.

character_errors = 0
total_characters = 0


for actual, predicted in zip(
    actual_words,
    predicted_words
):

    character_errors += edit_distance(
        actual,
        predicted
    )

    total_characters += len(actual)


cer = character_errors / max(
    total_characters,
    1
)


# ------------------------------------------------------------
# 8. DISPLAY ALL PERFORMANCE METRICS
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("FINAL PERFORMANCE METRICS")
print("=" * 60)

print(
    f"BLEU-1              : {bleu1:.4f}"
)

print(
    f"BLEU-2              : {bleu2:.4f}"
)

print(
    f"BLEU-3              : {bleu3:.4f}"
)

print(
    f"METEOR              : {meteor:.4f}"
)

print(
    f"ROUGE-1 Recall      : {rouge1:.4f}"
)

print(
    f"ROUGE-2 Recall      : {rouge2:.4f}"
)

print(
    f"ROUGE-3 Recall      : {rouge3:.4f}"
)

print(
    f"WER                 : {wer:.4f}"
)

print(
    f"CER                 : {cer:.4f}"
)


# ------------------------------------------------------------
# 9. BAR GRAPH - QUALITY METRICS
# ------------------------------------------------------------

quality_names = [
    "BLEU-1",
    "BLEU-2",
    "BLEU-3",
    "METEOR",
    "ROUGE-1",
    "ROUGE-2",
    "ROUGE-3"
]

quality_values = [
    bleu1,
    bleu2,
    bleu3,
    meteor,
    rouge1,
    rouge2,
    rouge3
]


plt.figure(figsize=(11, 6))

bars = plt.bar(
    quality_names,
    quality_values
)

plt.title(
    "NLP Performance Metrics"
)

plt.xlabel(
    "Metrics"
)

plt.ylabel(
    "Score"
)

plt.ylim(
    0,
    max(
        1.0,
        max(quality_values) * 1.15
    )
)


for bar, value in zip(
    bars,
    quality_values
):

    plt.text(
        bar.get_x() +
        bar.get_width() / 2,

        bar.get_height() + 0.01,

        f"{value:.3f}",

        ha="center"
    )


plt.xticks(rotation=20)

plt.tight_layout()

plt.show()


# ------------------------------------------------------------
# 10. BAR GRAPH - ERROR METRICS
# ------------------------------------------------------------

error_names = [
    "WER",
    "CER"
]

error_values = [
    wer,
    cer
]


plt.figure(figsize=(7, 5))

bars = plt.bar(
    error_names,
    error_values
)

plt.title(
    "Error Metrics"
)

plt.xlabel(
    "Metrics"
)

plt.ylabel(
    "Error Rate"
)


for bar, value in zip(
    bars,
    error_values
):

    plt.text(
        bar.get_x() +
        bar.get_width() / 2,

        bar.get_height() + 0.01,

        f"{value:.3f}",

        ha="center"
    )


plt.tight_layout()

plt.show()


print("\nPerformance evaluation completed successfully.")
```
