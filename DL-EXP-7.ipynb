```python
# ============================================================
# DL EXPERIMENT 7
# LSTM FOR NEXT-WORD PREDICTION
# Dataset: 3 Project Gutenberg Books
# ============================================================

import re
import random
import warnings
import numpy as np
import matplotlib.pyplot as plt
import requests
import nltk
import tensorflow as tf

from nltk import word_tokenize, pos_tag
from nltk.corpus import wordnet
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

warnings.filterwarnings("ignore")

# ------------------------------------------------------------
# 1. SET RANDOM SEEDS
# ------------------------------------------------------------

np.random.seed(42)
random.seed(42)
tf.random.set_seed(42)

print("=" * 60)
print("LSTM NEXT-WORD PREDICTION")
print("=" * 60)

# ------------------------------------------------------------
# 2. DOWNLOAD NLTK PACKAGES
# ------------------------------------------------------------

for package in [
    "punkt",
    "punkt_tab",
    "averaged_perceptron_tagger_eng",
    "wordnet"
]:
    nltk.download(package, quiet=True)

# ------------------------------------------------------------
# 3. DOWNLOAD THREE BOOKS
# ------------------------------------------------------------

book_ids = {
    11: "Alice's Adventures in Wonderland",
    84: "Frankenstein",
    1661: "The Adventures of Sherlock Holmes"
}

all_books = []
loaded_books = []

print("\nDOWNLOADING DATASET")
print("-" * 50)

for book_id, book_title in book_ids.items():
    try:
        url = (
            f"https://www.gutenberg.org/cache/epub/"
            f"{book_id}/pg{book_id}.txt"
        )

        response = requests.get(url, timeout=60)
        response.raise_for_status()
        book_text = response.text

        start = re.search(
            r"\*\*\* START OF (?:THE|THIS) PROJECT GUTENBERG EBOOK.*?\*\*\*",
            book_text,
            flags=re.I | re.S
        )

        end = re.search(
            r"\*\*\* END OF (?:THE|THIS) PROJECT GUTENBERG EBOOK",
            book_text,
            flags=re.I
        )

        if start:
            book_text = book_text[start.end():]

        if end:
            book_text = re.split(
                r"\*\*\* END OF (?:THE|THIS) PROJECT GUTENBERG EBOOK",
                book_text,
                flags=re.I
            )[0]

        all_books.append(book_text)
        loaded_books.append(book_title)
        print("Loaded:", book_title)

    except Exception as e:
        print("Skipped:", book_title, "|", e)

if not all_books:
    raise RuntimeError(
        "Books could not be downloaded. Check your internet connection."
    )

raw_text = "\n".join(all_books)

print("\nBOOKS LOADED:", len(loaded_books))
print("BOOK TITLES:", loaded_books)
print("TOTAL CHARACTERS:", len(raw_text))
print("\nORIGINAL TEXT:\n", raw_text[:200])

# ------------------------------------------------------------
# 4. TEXT PREPROCESSING AND CLEANING
# ------------------------------------------------------------

text = raw_text.lower()
text = re.sub(r"[^a-z\s']", " ", text)
text = re.sub(r"\s+", " ", text).strip()

print("\nCLEANED TEXT")
print(text[:200])

tokens = word_tokenize(text)

print("\nTOTAL WORD TOKENS:", len(tokens))
print("SAMPLE TOKENS:", tokens[:15])

# WordNet dictionary analysis
sample_words = [w for w in tokens[:50] if w.isalpha()]

dictionary_words = [
    w for w in sample_words if wordnet.synsets(w)
]

unknown_words = [
    w for w in sample_words if not wordnet.synsets(w)
]

print("\nWORDNET ANALYSIS")
print("Dictionary words:", dictionary_words[:10])
print("Words not found:", unknown_words[:10])

# Part-of-speech tagging
print("\nPART-OF-SPEECH TAGGING")
print(pos_tag(tokens[:15]))

# ------------------------------------------------------------
# 5. TOKENIZATION AND WORD EMBEDDINGS
# ------------------------------------------------------------

MAX_WORDS = 5000
SEQ_LEN = 8
EMBED_DIM = 32

tokenizer = Tokenizer(
    num_words=MAX_WORDS,
    oov_token="<OOV>"
)

tokenizer.fit_on_texts([text])

vocab_size = min(
    MAX_WORDS,
    len(tokenizer.word_index) + 1
)

word_ids = tokenizer.texts_to_sequences([text])[0]

print("\nTOKENIZATION DETAILS")
print("Vocabulary size:", vocab_size)
print("Sequence length:", SEQ_LEN)
print("Embedding dimensions:", EMBED_DIM)

print("\nWORD-TO-INDEX MAPPING")
print(list(tokenizer.word_index.items())[:15])

# ------------------------------------------------------------
# 6. CREATE SEQUENCES
# ------------------------------------------------------------

n = len(word_ids)

split1 = int(n * 0.80)
split2 = int(n * 0.90)

train_ids = word_ids[:split1]
val_ids = word_ids[split1:split2]
test_ids = word_ids[split2:]


def create_sequences(ids, seq_len):
    X, y = [], []

    for i in range(seq_len, len(ids)):
        X.append(ids[i - seq_len:i])
        y.append(ids[i])

    return (
        np.asarray(X, dtype=np.int32),
        np.asarray(y, dtype=np.int32)
    )


X_train, y_train = create_sequences(train_ids, SEQ_LEN)
X_val, y_val = create_sequences(val_ids, SEQ_LEN)
X_test, y_test = create_sequences(test_ids, SEQ_LEN)

print("\nSEQUENCE DATA")
print("X_train:", X_train.shape)
print("y_train:", y_train.shape)
print("X_val:", X_val.shape)
print("y_val:", y_val.shape)
print("X_test:", X_test.shape)
print("y_test:", y_test.shape)

id_to_word = {
    index: word
    for word, index in tokenizer.word_index.items()
    if index < vocab_size
}


def decode_sequence(sequence):
    return [
        id_to_word.get(int(i), "?")
        for i in sequence
    ]


print("\nSAMPLE INPUT WORDS:", decode_sequence(X_train[0]))
print("INPUT TOKEN IDS:", X_train[0])
print("ACTUAL NEXT WORD:", id_to_word.get(int(y_train[0]), "?"))

# ------------------------------------------------------------
# 7. BUILD AND COMPILE THE LSTM MODEL
# ------------------------------------------------------------

model = Sequential([
    tf.keras.Input(shape=(SEQ_LEN,)),

    Embedding(
        input_dim=vocab_size,
        output_dim=EMBED_DIM,
        mask_zero=True,
        name="Word_Embedding"
    ),

    LSTM(32, name="LSTM_Layer"),

    Dropout(0.2),

    Dense(32, activation="relu"),

    Dense(
        vocab_size,
        activation="softmax",
        name="Next_Word_Output"
    )
])

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=[
        "accuracy",
        tf.keras.metrics.SparseTopKCategoricalAccuracy(
            k=3,
            name="top3_accuracy"
        )
    ]
)

print("\nMODEL SUMMARY")
model.summary()

# ------------------------------------------------------------
# 8. TRAIN THE MODEL
# ------------------------------------------------------------

early_stop = EarlyStopping(
    monitor="val_loss",
    patience=10,
    restore_best_weights=True
)

print("\nTRAINING LSTM MODEL")
print("Maximum epochs: 30")
print("Early stopping patience: 10")

history = model.fit(
    X_train,
    y_train,
    validation_data=(X_val, y_val),
    epochs=30,
    batch_size=128,
    callbacks=[early_stop],
    shuffle=False,
    verbose=1
)

print("\nTraining completed.")
print("Epochs completed:", len(history.history["loss"]))

# ------------------------------------------------------------
# 9. MODEL PERFORMANCE
# ------------------------------------------------------------

test_loss, test_accuracy, test_top3 = model.evaluate(
    X_test,
    y_test,
    verbose=0
)

print("\nFINAL TEST PERFORMANCE")
print("=" * 50)
print(f"Test Loss      : {test_loss:.4f}")
print(f"Test Accuracy  : {test_accuracy * 100:.2f}%")
print(f"Top-3 Accuracy : {test_top3 * 100:.2f}%")

# ------------------------------------------------------------
# 10. ACCURACY AND LOSS GRAPHS
# ------------------------------------------------------------

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.plot(history.history["accuracy"], label="Training")
plt.plot(history.history["val_accuracy"], label="Validation")
plt.title("LSTM Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(alpha=0.3)

plt.subplot(1, 2, 2)
plt.plot(history.history["loss"], label="Training")
plt.plot(history.history["val_loss"], label="Validation")
plt.title("LSTM Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 11. WORD EMBEDDING VECTORS
# ------------------------------------------------------------

embedding_matrix = model.get_layer(
    "Word_Embedding"
).get_weights()[0]

print("\nWORD EMBEDDING MATRIX")
print("Embedding matrix shape:", embedding_matrix.shape)

for word in ["alice", "rabbit", "queen", "king", "said"]:
    word_id = tokenizer.word_index.get(word)

    if word_id is not None and word_id < vocab_size:
        print("\nWord:", word)
        print("Token ID:", word_id)
        print(
            "First 10 vector values:",
            np.round(embedding_matrix[word_id][:10], 4)
        )

print("\nFIRST FIVE VECTORS")
print(np.round(embedding_matrix[1:6, :8], 4))

# ------------------------------------------------------------
# 12. NEXT-WORD PREDICTION
# ------------------------------------------------------------

def predict_next_word(input_text, top_n=5, check_answer=False):

    cleaned = input_text.lower()
    cleaned = re.sub(r"[^a-z\s']", " ", cleaned)
    cleaned = re.sub(r"\s+", " ", cleaned).strip()

    sequence = tokenizer.texts_to_sequences([cleaned])[0]
    sequence = sequence[-SEQ_LEN:]

    if not sequence:
        print("Please enter valid words.")
        return

    if all(word_id == 1 for word_id in sequence):
        print("Input words are outside the training vocabulary.")
        print("Try words used in the training books.")
        return

    padded = pad_sequences(
        [sequence],
        maxlen=SEQ_LEN,
        padding="pre"
    )

    probabilities = model.predict(
        padded,
        verbose=0
    )[0].copy()

    # Exclude padding and unknown-word tokens
    probabilities[0] = 0
    probabilities[1] = 0

    valid_ids = [
        i for i in range(2, len(probabilities))
        if i in id_to_word
    ]

    top_ids = sorted(
        valid_ids,
        key=lambda i: probabilities[i],
        reverse=True
    )[:top_n]

    if not top_ids:
        print("No valid prediction found.")
        return

    print("\nINPUT:", input_text)
    print("ENCODED INPUT:", padded[0])
    print("\nTOP NEXT-WORD PREDICTIONS")
    print("-" * 45)

    for rank, word_id in enumerate(top_ids, start=1):
        word = id_to_word[word_id]
        probability = probabilities[word_id] * 100

        print(
            f"{rank}. {word:<18}"
            f"Probability: {probability:.2f}%"
        )

    predicted_word = id_to_word[top_ids[0]]

    print("\nPREDICTED NEXT WORD:", predicted_word)
    print("COMPLETED TEXT:", input_text + " " + predicted_word)

    # Check against the actual next word
    if check_answer:
        actual_word = input(
            "\nEnter the actual next word "
            "(press Enter to skip): "
        ).strip()

        if actual_word:
            actual_word = re.sub(
                r"[^a-z']",
                "",
                actual_word.lower()
            )

            print("\nCORRECTNESS CHECK")
            print("Predicted word:", predicted_word)
            print("Actual word:", actual_word)

            if predicted_word.lower() == actual_word:
                print("RESULT: CORRECT")
            else:
                print("RESULT: WRONG")
        else:
            print("Correctness check skipped.")

# ------------------------------------------------------------
# 13. SAMPLE PREDICTIONS
# ------------------------------------------------------------

print("\nSAMPLE PREDICTIONS")

predict_next_word("alice was")
predict_next_word("the white rabbit")
predict_next_word("the queen")

# ------------------------------------------------------------
# 14. INTERACTIVE USER INPUT
# Type exit to stop
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("INTERACTIVE NEXT-WORD PREDICTION")
print("Enter a sentence or type 'exit' to stop.")
print("=" * 60)

while True:

    user_text = input("\nEnter a sentence: ").strip()

    if user_text.lower() == "exit":
        print("Prediction stopped. Experiment completed!")
        break

    if not user_text:
        print("Please enter a sentence.")
        continue

    predict_next_word(
        user_text,
        top_n=5,
        check_answer=True
    )

print("\nDL EXPERIMENT 7 FINISHED SUCCESSFULLY")
```
