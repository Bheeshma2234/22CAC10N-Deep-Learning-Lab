```python
# ============================================================
# EXPERIMENT-2: MLP FOR BREAST CANCER DATASET
# Complete Implementation Pipeline with All Visualizations
# ============================================================

# ------------------------------------------------------------
# STEP 1: Import required Python libraries
# ------------------------------------------------------------
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf

from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.feature_selection import SelectKBest, f_classif, f_regression
from sklearn.metrics import (accuracy_score, confusion_matrix,
                             classification_report, ConfusionMatrixDisplay,
                             roc_curve, auc,
                             mean_squared_error, mean_absolute_error, r2_score)

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input

print("="*70)
print("EXPERIMENT-2 : MLP FOR CLASSIFICATION & REGRESSION")
print("="*70)

# ============================================================
# PART A: CLASSIFICATION TASK (Predicting Malignant vs Benign)
# ============================================================
print("\n" + "="*50)
print("PART A: CLASSIFICATION TASK")
print("="*50)

# ------------------------------------------------------------
# STEP 2: Load the Breast Cancer Wisconsin dataset
# ------------------------------------------------------------
data = load_breast_cancer()
X = pd.DataFrame(data.data, columns=data.feature_names)
y_class = pd.Series(data.target)

# ------------------------------------------------------------
# STEP 3: Perform exploratory data analysis (EDA)
# ------------------------------------------------------------
print("\n--- Dataset Overview ---")
print("Dataset Shape:", X.shape)
print("\nDataset Info:")
print(X.info())
print("\nStatistical Summary:\n", X.describe())

# Check Missing Values
print("\nMissing Values Count:\n", X.isnull().sum())

# EDA Graph 1: Class Distribution
plt.figure(figsize=(5, 4))
y_class.value_counts().plot(kind="bar", color=['teal', 'crimson'])
plt.title("Class Distribution (0 = Malignant, 1 = Benign)")
plt.xlabel("Class")
plt.ylabel("Count")
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()

# EDA Graph 2: All Feature Histograms
X.hist(figsize=(18, 15), color='steelblue', edgecolor='black')
plt.suptitle("Feature Histograms", fontsize=16)
plt.tight_layout()
plt.show()

# EDA Graph 3: Boxplots BEFORE Outlier Capping
plt.figure(figsize=(18, 6))
X.boxplot(rot=90)
plt.title("Feature Boxplots (BEFORE Outlier Capping)")
plt.tight_layout()
plt.show()

# EDA Graph 4: NORMAL / RAW CORRELATION HEATMAP (Before Preprocessing)
plt.figure(figsize=(12, 10))
corr_raw = X.corr()
plt.imshow(corr_raw, cmap="coolwarm", aspect="auto")
plt.colorbar()
plt.title("Normal Correlation Heatmap (Raw Data - 30 Features)")
plt.show()

# EDA Graph 5: Feature Scatter Plot
plt.figure(figsize=(6, 5))
plt.scatter(X.iloc[:, 0], X.iloc[:, 1], c=y_class, cmap="coolwarm", alpha=0.8)
plt.xlabel(X.columns[0])
plt.ylabel(X.columns[1])
plt.title("Scatter Plot (Mean Radius vs Mean Texture)")
plt.grid(True)
plt.show()

# ------------------------------------------------------------
# STEP 4: Preprocess, Impute, Handle Outliers, Select & Standardize
# ------------------------------------------------------------

# 4a. Explicit Missing Value Imputation
imputer = SimpleImputer(strategy='mean')
X_imputed = pd.DataFrame(imputer.fit_transform(X), columns=X.columns)

# 4b. Outlier Capping (IQR Method)
X_clean = X_imputed.copy()
for col in X_clean.columns:
    Q1 = X_clean[col].quantile(0.25)
    Q3 = X_clean[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    X_clean[col] = np.clip(X_clean[col], lower_bound, upper_bound)

print("\nOutliers successfully handled using IQR capping.")

# EDA Graph 6: Boxplots AFTER Outlier Capping
plt.figure(figsize=(18, 6))
X_clean.boxplot(rot=90)
plt.title("Feature Boxplots (AFTER Outlier Capping)")
plt.tight_layout()
plt.show()

# 4c. Feature Selection (Top 15 Features via ANOVA F-test)
k_features_class = 15
selector_class = SelectKBest(score_func=f_classif, k=k_features_class)
X_selected_class = selector_class.fit_transform(X_clean, y_class)

selected_cols_class = X_clean.columns[selector_class.get_support()]
print(f"\nTop {k_features_class} Selected Features for Classification:\n", list(selected_cols_class))

# EDA Graph 7: Classification Feature Importance Bar Plot
plt.figure(figsize=(10, 4))
feature_scores_class = pd.Series(selector_class.scores_, index=X_clean.columns)
feature_scores_class.nlargest(15).plot(kind='bar', color='teal')
plt.title("Top 15 Features by ANOVA F-score (Classification)")
plt.ylabel("F-score")
plt.grid(axis='y', linestyle='--')
plt.show()

# EDA Graph 8: CORRELATION HEATMAP OF SELECTED FEATURES (Post-Feature Selection)
plt.figure(figsize=(10, 8))
corr_selected = X_clean[selected_cols_class].corr()
plt.imshow(corr_selected, cmap="coolwarm", aspect="auto")
plt.colorbar()
plt.xticks(range(len(selected_cols_class)), selected_cols_class, rotation=90)
plt.yticks(range(len(selected_cols_class)), selected_cols_class)
plt.title("Correlation Heatmap (Selected 15 Features)")
plt.tight_layout()
plt.show()

# 4d. Standardize Features
scaler_class = StandardScaler()
X_scaled_class = scaler_class.fit_transform(X_selected_class)

# ------------------------------------------------------------
# STEP 5: Split the data into training and testing sets
# ------------------------------------------------------------
X_train_c, X_test_c, y_train_c, y_test_c = train_test_split(
    X_scaled_class, y_class, test_size=0.2, random_state=42, stratify=y_class
)

# ------------------------------------------------------------
# STEP 6: Build the MLP model using TensorFlow/Keras
# ------------------------------------------------------------
model_class = Sequential([
    Input(shape=(X_scaled_class.shape[1],)),
    Dense(16, activation='relu'),
    Dense(8, activation='relu'),
    Dense(1, activation='sigmoid')
])

print("\nClassification Model Summary:")
model_class.summary()

model_class.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# ------------------------------------------------------------
# STEP 7: Train the network
# ------------------------------------------------------------
history_class = model_class.fit(
    X_train_c, y_train_c,
    epochs=50,
    batch_size=16,
    validation_split=0.2,
    verbose=1
)

# ------------------------------------------------------------
# STEP 8: Predict the test samples
# ------------------------------------------------------------
prob_class = model_class.predict(X_test_c)
pred_class = (prob_class > 0.5).astype(int)

# ------------------------------------------------------------
# STEP 9: Evaluate the model using performance metrics
# ------------------------------------------------------------
print("\n--- CLASSIFICATION METRICS ---")
print("Accuracy Score:", accuracy_score(y_test_c, pred_class))
cm_class = confusion_matrix(y_test_c, pred_class)
print("\nConfusion Matrix:\n", cm_class)
print("\nClassification Report:\n", classification_report(y_test_c, pred_class, target_names=data.target_names))

# ------------------------------------------------------------
# STEP 10: Visualize learning curves and confusion matrix
# ------------------------------------------------------------
# Graph 9: Classification Accuracy Curve
plt.figure(figsize=(6, 4))
plt.plot(history_class.history['accuracy'], label='Train Accuracy')
plt.plot(history_class.history['val_accuracy'], label='Validation Accuracy')
plt.title("Classification Accuracy Curve")
plt.xlabel("Epochs")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(True)
plt.show()

# Graph 10: Classification Loss Curve
plt.figure(figsize=(6, 4))
plt.plot(history_class.history['loss'], label='Train Loss')
plt.plot(history_class.history['val_loss'], label='Validation Loss')
plt.title("Classification Loss Curve")
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)
plt.show()

# Graph 11: Confusion Matrix Display
ConfusionMatrixDisplay(cm_class, display_labels=data.target_names).plot(cmap='Blues')
plt.title("Classification Confusion Matrix")
plt.show()

# Graph 12: ROC Curve
fpr, tpr, _ = roc_curve(y_test_c, prob_class)
roc_auc = auc(fpr, tpr)
plt.figure(figsize=(6, 4))
plt.plot(fpr, tpr, label=f"AUC = {roc_auc:.3f}", color='darkorange', lw=2)
plt.plot([0, 1], [0, 1], 'k--')
plt.title("ROC Curve")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.legend()
plt.grid(True)
plt.show()


# ============================================================
# PART B: REGRESSION TASK (Predicting Continuous Target: Mean Radius)
# ============================================================
print("\n" + "="*50)
print("PART B: REGRESSION TASK")
print("="*50)

# ------------------------------------------------------------
# STEP 2 & 3: Define Regression Target and Features
# ------------------------------------------------------------
y_reg = X_clean['mean radius']
X_reg = X_clean.drop(columns=['mean radius'])

# ------------------------------------------------------------
# STEP 4: Feature Selection & Scaling for Regression
# ------------------------------------------------------------
k_features_reg = 15
selector_reg = SelectKBest(score_func=f_regression, k=k_features_reg)
X_selected_reg = selector_reg.fit_transform(X_reg, y_reg)

selected_cols_reg = X_reg.columns[selector_reg.get_support()]
print(f"\nTop {k_features_reg} Selected Features for Regression:\n", list(selected_cols_reg))

# Graph 13: Regression Feature Importance Bar Plot
plt.figure(figsize=(10, 4))
feature_scores_reg = pd.Series(selector_reg.scores_, index=X_reg.columns)
feature_scores_reg.nlargest(15).plot(kind='bar', color='coral')
plt.title("Top 15 Features by F-regression Score (Regression)")
plt.ylabel("F-score")
plt.grid(axis='y', linestyle='--')
plt.show()

scaler_reg = StandardScaler()
X_scaled_reg = scaler_reg.fit_transform(X_selected_reg)

# ------------------------------------------------------------
# STEP 5: Split the data into training and testing sets
# ------------------------------------------------------------
X_train_r, X_test_r, y_train_r, y_test_r = train_test_split(
    X_scaled_reg, y_reg, test_size=0.2, random_state=42
)

# ------------------------------------------------------------
# STEP 6: Build the MLP Regression Model
# ------------------------------------------------------------
model_reg = Sequential([
    Input(shape=(X_scaled_reg.shape[1],)),
    Dense(32, activation='relu'),
    Dense(16, activation='relu'),
    Dense(1, activation='linear')
])

print("\nRegression Model Summary:")
model_reg.summary()

model_reg.compile(
    optimizer='adam',
    loss='mean_squared_error',
    metrics=['mae']
)

# ------------------------------------------------------------
# STEP 7: Train the network
# ------------------------------------------------------------
history_reg = model_reg.fit(
    X_train_r, y_train_r,
    epochs=50,
    batch_size=16,
    validation_split=0.2,
    verbose=1
)

# ------------------------------------------------------------
# STEP 8: Predict the test samples
# ------------------------------------------------------------
pred_reg = model_reg.predict(X_test_r).flatten()

# ------------------------------------------------------------
# STEP 9: Evaluate the model using regression performance metrics (Errors)
# ------------------------------------------------------------
mse = mean_squared_error(y_test_r, pred_reg)
rmse = np.sqrt(mse)
mae = mean_absolute_error(y_test_r, pred_reg)
r2 = r2_score(y_test_r, pred_reg)

print("\n--- REGRESSION ERROR EVALUATION METRICS ---")
print(f"Mean Squared Error (MSE)      : {mse:.4f}")
print(f"Root Mean Squared Error (RMSE) : {rmse:.4f}")
print(f"Mean Absolute Error (MAE)     : {mae:.4f}")
print(f"R-squared Score (R²)          : {r2:.4f}")

# ------------------------------------------------------------
# STEP 10: Visualize learning curves and regression predictions
# ------------------------------------------------------------
# Graph 14: Regression Training Loss Curve (MSE)
plt.figure(figsize=(6, 4))
plt.plot(history_reg.history['loss'], label='Train Loss (MSE)')
plt.plot(history_reg.history['val_loss'], label='Validation Loss (MSE)')
plt.title("Regression Training Loss Curve (MSE)")
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)
plt.show()

# Graph 15: Regression MAE Metric Curve
plt.figure(figsize=(6, 4))
plt.plot(history_reg.history['mae'], label='Train MAE', color='green')
plt.plot(history_reg.history['val_mae'], label='Validation MAE', color='orange')
plt.title("Regression MAE Curve")
plt.xlabel("Epochs")
plt.ylabel("Mean Absolute Error")
plt.legend()
plt.grid(True)
plt.show()

# Graph 16: Actual vs Predicted Scatter Plot
plt.figure(figsize=(6, 5))
plt.scatter(y_test_r, pred_reg, alpha=0.7, color='purple', edgecolors='k')
plt.plot([y_test_r.min(), y_test_r.max()], [y_test_r.min(), y_test_r.max()], 'r--', lw=2, label='Ideal Fit Line')
plt.xlabel("Actual Mean Radius")
plt.ylabel("Predicted Mean Radius")
plt.title(f"Actual vs Predicted Mean Radius (R² = {r2:.3f})")
plt.legend()
plt.grid(True)
plt.show()

print("\nExperiment 2 Completed Successfully!")
```
